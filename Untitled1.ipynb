```python
import numpy as np
import pandas as pd
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt

# --- 1. CONFIGURATION ---
N_STUDENTS = 100  # Number of virtual students
N_QUESTIONS_TOTAL = 20 # Total questions in the bank
N_QUESTIONS_SESSION = 15 # Number of questions each student will answer
SEED = 42 # For reproducibility

# --- 2. IRT MODEL FUNCTIONS (2-Parameter Logistic Model) ---
def sigmoid(x):
    """The logistic function."""
    return 1 / (1 + np.exp(-x))

def prob_correct(theta, a, b):
    """
    Calculates the probability of a correct answer using the 2-PL IRT model.
    theta: Student's ability level
    a: Question's discrimination parameter
    b: Question's difficulty parameter
    """
    return sigmoid(a * (theta - b))

# --- 3. SIMULATION SETUP ---
def run_simulation():
    """Main function to run the entire simulation."""
    np.random.seed(SEED)

    # Load question bank
    try:
        questions_df = pd.read_csv('questions.csv')
        print("Successfully loaded questions.csv")
    except FileNotFoundError:
        print("Error: questions.csv not found. Please ensure the file is in the same directory.")
        return None

    # Generate virtual students with a normal distribution of ability levels (theta)
    student_abilities = np.random.normal(loc=0.0, scale=1.0, size=N_STUDENTS)
    print(f"Generated {N_STUDENTS} virtual students.")

    # --- 4. RUN EXPERIMENTAL & CONTROL GROUPS ---
    results_control = []
    results_experimental = []

    print("\nRunning simulations for Control and Experimental groups...")
    for i in range(N_STUDENTS):
        theta = student_abilities[i]
        
        # --- Control Group: Random Question Selection ---
        score_control = 0
        answered_indices_control = np.random.choice(N_QUESTIONS_TOTAL, N_QUESTIONS_SESSION, replace=False)
        for q_idx in answered_indices_control:
            a = questions_df.loc[q_idx, 'initial_discrimination_a']
            b = questions_df.loc[q_idx, 'initial_difficulty_b']
            
            # Simulate student response
            if np.random.rand() < prob_correct(theta, a, b):
                score_control += 1
        results_control.append(score_control / N_QUESTIONS_SESSION)

        # --- Experimental Group: IRT-based Personalized Selection ---
        score_experimental = 0
        theta_estimate = 0.0 # Start with an average ability estimate
        
        available_indices = list(range(N_QUESTIONS_TOTAL))
        
        for _ in range(N_QUESTIONS_SESSION):
            # Select the most informative question
            # (The one where P(correct) is closest to 0.75 for the current ability estimate)
            question_info = []
            for q_idx in available_indices:
                a = questions_df.loc[q_idx, 'initial_discrimination_a']
                b = questions_df.loc[q_idx, 'initial_difficulty_b']
                p = prob_correct(theta_estimate, a, b)
                info = abs(p - 0.75) # A simple proxy for information gain
                question_info.append((info, q_idx))
            
            best_q_idx = min(question_info, key=lambda x: x[0])[1]
            available_indices.remove(best_q_idx)

            # Simulate student response
            a = questions_df.loc[best_q_idx, 'initial_discrimination_a']
            b = questions_df.loc[best_q_idx, 'initial_difficulty_b']
            correct = np.random.rand() < prob_correct(theta, a, b)
            
            if correct:
                score_experimental += 1
                # Simple Bayesian update: if correct, increase ability estimate
                theta_estimate += 0.25 
            else:
                # If incorrect, decrease ability estimate
                theta_estimate -= 0.25
        
        results_experimental.append(score_experimental / N_QUESTIONS_SESSION)

    print("Simulations complete.")
    return results_control, results_experimental

# --- 5. ANALYZE AND REPORT RESULTS ---
def analyze_and_plot(control, experimental):
    """Calculates statistics and generates the plot."""
    
    # Calculate means and standard deviations
    mean_control = np.mean(control)
    std_control = np.std(control)
    mean_experimental = np.mean(experimental)
    std_experimental = np.std(experimental)

    # Perform independent t-test
    t_stat, p_value = ttest_ind(experimental, control)

    # --- Generate Plot ---
    labels = ['Control Group (Random)', 'Experimental Group (IRT-Personalized)']
    means = [mean_control, mean_experimental]
    std_devs = [std_control, std_experimental]
    
    fig, ax = plt.subplots()
    ax.bar(labels, means, yerr=std_devs, capsize=5, color=['skyblue', 'lightgreen'])
    ax.set_ylabel('Average Score')
    ax.set_title('Comparison of Student Performance')
    ax.set_ylim(0, 1.0)
    plt.tight_layout()
    
    # Save the figure
    plt.savefig('performance_comparison.png')
    print("\nGenerated plot: performance_comparison.png")

    # --- Print Results for Me ---
    print("\n--- RESULTS FOR AI AUTHOR ---")
    print(f"Control Group Mean Score: {mean_control:.4f}")
    print(f"Control Group Std Dev: {std_control:.4f}")
    print(f"Experimental Group Mean Score: {mean_experimental:.4f}")
    print(f"Experimental Group Std Dev: {std_experimental:.4f}")
    print(f"Performance Improvement: {(mean_experimental - mean_control) / mean_control * 100:.2f}%")
    print(f"T-statistic: {t_stat:.4f}")
    print(f"P-value: {p_value:.4f}")
    print("-----------------------------\n")

# --- MAIN EXECUTION ---
if __name__ == "__main__":
    results = run_simulation()
    if results:
        analyze_and_plot(results[0], results[1])
```

```python
import numpy as np
import pandas as pd
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt
from scipy.optimize import minimize
# --- 1. CONFIGURATION ---
N_STUDENTS = 10000 # Number of virtual students
N_QUESTIONS_TOTAL = 20 # Total questions in the bank
N_QUESTIONS_SESSION = 15 # Number of questions each student will answer
SEED = 42 # For reproducibility
# --- 2. IRT MODEL FUNCTIONS (2-Parameter Logistic Model) ---
def sigmoid(x):
    """The logistic function."""
    # Clip to avoid overflow in np.exp
    x = np.clip(x, -100, 100)
    return 1 / (1 + np.exp(-x))
def prob_correct(theta, a, b):
    """Calculates the probability of a correct answer using the 2-PL IRT model."""
    return sigmoid(a * (theta - b))
# --- NEW: Fisher Information and Likelihood for better adaptive logic ---
def fisher_information(theta, a, b):
    """Calculates the Fisher Information for a given item at an ability level theta."""
    p = prob_correct(theta, a, b)
    return (a**2) * p * (1 - p)
def negative_log_likelihood(theta, responses, questions_df):
    """Negative log-likelihood function for MLE ability estimation."""
    log_likelihood = 0.0
    for q_idx, response in responses.items():
        a = questions_df.loc[q_idx, 'initial_discrimination_a']
        b = questions_df.loc[q_idx, 'initial_difficulty_b']
        p = prob_correct(theta, a, b)
        # Add a small epsilon to avoid log(0)
        p = np.clip(p, 1e-9, 1 - 1e-9)
        if response == 1:
            log_likelihood += np.log(p)
        else:
            log_likelihood += np.log(1 - p)
    return -log_likelihood
# --- 3. SIMULATION SETUP ---
def run_simulation():
    """Main function to run the entire simulation with all four groups."""
    np.random.seed(SEED)
    try:
        questions_df = pd.read_csv('questions_expanded.csv')
        print("Successfully loaded questions_expanded.csv")
    except FileNotFoundError:
        print("Error: questions.csv not found.")
        return None
    student_abilities = np.random.normal(loc=0.0, scale=1.0, size=N_STUDENTS)
    print(f"Generated {N_STUDENTS} virtual students.")
    # --- NEW: Prepare lists for all four experimental groups ---
    results_control_random = []
    results_control_fixed = []
    results_experimental_adaptive = []
    results_experimental_perfect = []
    # --- NEW: Get indices for the fixed-difficulty baseline ---
    # In your paper, "Medium" difficulty maps to b=0.0 [cite: 119]
    medium_difficulty_indices = questions_df[questions_df['initial_difficulty_b'] == 0.0].index.tolist()
    if len(medium_difficulty_indices) < N_QUESTIONS_SESSION:
        print(f"Warning: Not enough medium difficulty questions for the session (found {len(medium_difficulty_indices)}, need {N_QUESTIONS_SESSION}).")
    print("\nRunning simulations for all 4 groups...")
    for i in range(N_STUDENTS):
        theta_true = student_abilities[i]
        # --- Group 1: Control (Random Selection) ---
        score_control_random = 0
        q_indices = np.random.choice(N_QUESTIONS_TOTAL, N_QUESTIONS_SESSION, replace=False)
        for q_idx in q_indices:
            a = questions_df.loc[q_idx, 'initial_discrimination_a']
            b = questions_df.loc[q_idx, 'initial_difficulty_b']
            if np.random.rand() < prob_correct(theta_true, a, b):
                score_control_random += 1
        results_control_random.append(score_control_random / N_QUESTIONS_SESSION)
        # --- NEW - Group 2: Control (Fixed-Difficulty Selection) ---
        score_control_fixed = 0
        if len(medium_difficulty_indices) >= N_QUESTIONS_SESSION:
            q_indices = np.random.choice(medium_difficulty_indices, N_QUESTIONS_SESSION, replace=False)
            for q_idx in q_indices:
                a = questions_df.loc[q_idx, 'initial_discrimination_a']
                b = questions_df.loc[q_idx, 'initial_difficulty_b']
                if np.random.rand() < prob_correct(theta_true, a, b):
                    score_control_fixed += 1
            results_control_fixed.append(score_control_fixed / N_QUESTIONS_SESSION)
        else: # Handle case where there aren't enough questions
            results_control_fixed.append(np.nan)
        # --- Group 3: Experimental (IRT-Personalized Adaptive) ---
        # REVERT THIS BLOCK TO YOUR ORIGINAL, SIMPLER LOGIC
        score_experimental_adaptive = 0
        theta_estimate = 0.0 # Start with an average ability estimate
        available_indices = list(range(N_QUESTIONS_TOTAL))
        for _ in range(N_QUESTIONS_SESSION):
            # Select the most informative question
            # (Using the original, simpler heuristic)
            question_info = []
            for q_idx in available_indices:
                a = questions_df.loc[q_idx, 'initial_discrimination_a']
                b = questions_df.loc[q_idx, 'initial_difficulty_b']
                p = prob_correct(theta_estimate, a, b)
                # THIS IS YOUR ORIGINAL, FUNCTIONAL HEURISTIC
                info = abs(p - 0.75)
                question_info.append((info, q_idx))
            best_q_idx = min(question_info, key=lambda x: x[0])[1]
            available_indices.remove(best_q_idx)
            # Simulate student response
            a = questions_df.loc[best_q_idx, 'initial_discrimination_a']
            b = questions_df.loc[best_q_idx, 'initial_difficulty_b']
            correct = np.random.rand() < prob_correct(theta_true, a, b)
            if correct:
                score_experimental_adaptive += 1
                # THIS IS YOUR ORIGINAL, STABLE UPDATE RULE
                theta_estimate += 0.25
            else:
                theta_estimate -= 0.25
        results_experimental_adaptive.append(score_experimental_adaptive / N_QUESTIONS_SESSION)
        # --- NEW - Group 4: Experimental (Perfect Knowledge Selection) ---
        score_experimental_perfect = 0
        available_indices_perfect = list(range(N_QUESTIONS_TOTAL))
        for _ in range(N_QUESTIONS_SESSION):
            # Select the question with highest probability of correct (using true theta)
            question_info = []
            for q_idx in available_indices_perfect:
                a = questions_df.loc[q_idx, 'initial_discrimination_a']
                b = questions_df.loc[q_idx, 'initial_difficulty_b']
                p = prob_correct(theta_true, a, b)
                info = -p  # Negative to select max p via min
                question_info.append((info, q_idx))
            best_q_idx = min(question_info, key=lambda x: x[0])[1]
            available_indices_perfect.remove(best_q_idx)
            # Simulate student response
            a = questions_df.loc[best_q_idx, 'initial_discrimination_a']
            b = questions_df.loc[best_q_idx, 'initial_difficulty_b']
            correct = np.random.rand() < prob_correct(theta_true, a, b)
            if correct:
                score_experimental_perfect += 1
        results_experimental_perfect.append(score_experimental_perfect / N_QUESTIONS_SESSION)
    return student_abilities, results_control_random, results_control_fixed, results_experimental_adaptive, results_experimental_perfect
# --- 5. ANALYZE AND REPORT RESULTS ---
def analyze_and_plot(abilities, random, fixed, adaptive, perfect):
    """Calculates statistics, performs learner analysis, and generates the plot."""
   
    groups = {
        'Random': random,
        'Fixed-Difficulty': fixed,
        'IRT-Personalized': adaptive,
        'Perfect Knowledge': perfect
    }
   
    print("\n--- OVERALL RESULTS ---")
    for name, data in groups.items():
        mean_score = np.nanmean(data)
        std_dev = np.nanstd(data)
        print(f"{name} Group | Mean Score: {mean_score:.4f}, Std Dev: {std_dev:.4f}")
    # Perform t-tests against the adaptive group
    print("\n--- STATISTICAL TESTS (vs. IRT-Personalized) ---")
    t_stat_random, p_random = ttest_ind(adaptive, random)
    print(f"vs. Random: t-statistic={t_stat_random:.4f}, p-value={p_random:.4f}")
   
    t_stat_fixed, p_fixed = ttest_ind(adaptive, fixed, nan_policy='omit')
    print(f"vs. Fixed-Difficulty: t-statistic={t_stat_fixed:.4f}, p-value={p_fixed:.4f}")
    
    t_stat_perfect, p_perfect = ttest_ind(adaptive, perfect, nan_policy='omit')
    print(f"vs. Perfect Knowledge: t-statistic={t_stat_perfect:.4f}, p-value={p_perfect:.4f}")
    # --- NEW: Learner-Profile Analysis ---
    analysis_df = pd.DataFrame({
        'ability': abilities,
        'Random': random,
        'Fixed-Difficulty': fixed,
        'Adaptive': adaptive,
        'Perfect': perfect
    })
    analysis_df['ability_quartile'] = pd.qcut(analysis_df['ability'], 4, labels=['Q1 (Low)', 'Q2', 'Q3', 'Q4 (High)'])
   
    quartile_analysis = analysis_df.groupby('ability_quartile')[['Random', 'Fixed-Difficulty', 'Adaptive', 'Perfect']].mean()
    lifts = quartile_analysis.copy()
    lifts['Fixed Lift vs Random'] = quartile_analysis['Fixed-Difficulty'] - quartile_analysis['Random']
    lifts['Adaptive Lift vs Random'] = quartile_analysis['Adaptive'] - quartile_analysis['Random']
    lifts['Perfect Lift vs Random'] = quartile_analysis['Perfect'] - quartile_analysis['Random']
   
    print("\n--- LEARNER-PROFILE ANALYSIS (PERFORMANCE BY ABILITY QUARTILE) ---")
    print(quartile_analysis)
    print("\n--- PERFORMANCE LIFTS (vs. Random) ---")
    print(lifts[['Fixed Lift vs Random', 'Adaptive Lift vs Random', 'Perfect Lift vs Random']])
    print("----------------------------------------------------------------")
    # --- Generate Plot ---
    labels = list(groups.keys())
    means = [np.nanmean(d) for d in groups.values()]
    std_devs = [np.nanstd(d) for d in groups.values()]
   
    fig, ax = plt.subplots(figsize=(10, 6))
    colors = ['skyblue', 'orange', 'lightgreen', 'gold']
    ax.bar(labels, means, yerr=std_devs, capsize=5, color=colors)
    ax.set_ylabel('Average Score')
    ax.set_title('Comparison of Student Performance Across Four Conditions')
    ax.set_ylim(0, 1.0)
    plt.xticks(rotation=15, ha="right")
    plt.tight_layout()
   
    plt.savefig('performance_comparison.png')
    print("\nGenerated updated plot: performance_comparison.png")
# --- MAIN EXECUTION ---
if __name__ == "__main__":
    results = run_simulation()
    if results:
        analyze_and_plot(*results)
```
